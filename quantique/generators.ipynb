```python
def find_generators(p):
    """ finds all element g of set Zp* such that Zp* = {(g**n)%p for n in Zpm1} """
    Zp     = set(range(p))
    Zpm1   = set(range(p-1))
    generators    = []
    for v in list(Zp - {0}):
        list_arr = []
        for d in list(Zpm1):
            
            if (v**d) % p not in list_arr:
                list_arr.append((v**d) % p)
                #print (f"\t{v} generated {(v**d) % p}")
            else:
                print (f"\t{v} not a generator. breaking on {list_arr}")
                break
        if set(list_arr) == Zp - {0}:
            #print (f"\t{v} is a generator. {list_arr}")
            #print(f"{v}", list_arr)
            generators.append((v, list_arr))
    return generators
```

```python
def read_until_full_set(liste, pos):
    nb_val = len(liste)
    categories=list(set(liste))
    nb_categorie = len(categories)
    idx =  0
    while len(categories)!=0:
        v = liste[(idx+pos)%(nb_val)]
        if v in categories: categories.remove(v)
        idx += 1
    return idx
```

```python
import numpy as np

def trouver_coefficients_polynome_3(points):
    """
    Trouve les coefficients d'un polynôme de degré 2 qui passe par 3 points.
    
    Args:
        points (list): Une liste de 3 tuples, où chaque tuple est une coordonnée (x, y).
        
    Returns:
        numpy.ndarray: Un tableau contenant les coefficients [b, c, d].
    """
    # On prépare la matrice A du système d'équations Ax = B
    # Chaque ligne de la matrice correspond à un point (x, y)
    # et est de la forme [x³, x², x, 1]
    matrice_A = []
    for x, y in points:
        matrice_A.append([x**2, x, 1])
        
    # On prépare le vecteur B avec les valeurs de y
    vecteur_B = [p[1] for p in points]
    
    # On convertit les listes en tableaux NumPy
    A = np.array(matrice_A)
    B = np.array(vecteur_B)
    
    # On résout le système d'équations linéaires pour trouver les coefficients [a, b, c, d]
    coefficients = np.linalg.solve(A, B)
    
    return coefficients

def trouver_coefficients_polynome_2(points):
    """
    Trouve les coefficients d'un polynôme de degré 1 qui passe par 2 points.
    
    Args:
        points (list): Une liste de 2 tuples, où chaque tuple est une coordonnée (x, y).
        
    Returns:
        numpy.ndarray: Un tableau contenant les coefficients [c, d].
    """
    # On prépare la matrice A du système d'équations Ax = B
    # Chaque ligne de la matrice correspond à un point (x, y)
    # et est de la forme [ x, 1]
    matrice_A = []
    for x, y in points:
        matrice_A.append([x, 1])
        
    # On prépare le vecteur B avec les valeurs de y
    vecteur_B = [p[1] for p in points]
    
    # On convertit les listes en tableaux NumPy
    A = np.array(matrice_A)
    B = np.array(vecteur_B)
    
    # On résout le système d'équations linéaires pour trouver les coefficients [a, b, c, d]
    coefficients = np.linalg.solve(A, B)
    
    return coefficients
def trouver_coefficients_polynome_1(points):
    """
    Trouve les coefficients d'un polynôme de degré 1 qui passe par 2 points.
    
    Args:
        points (list): Une liste de 2 tuples, où chaque tuple est une coordonnée (x, y).
        
    Returns:
        numpy.ndarray: Un tableau contenant les coefficients [c, d].
    """
    # On prépare la matrice A du système d'équations Ax = B
    # Chaque ligne de la matrice correspond à un point (x, y)
    # et est de la forme [ x, 1]
    matrice_A = []
    for x, y in points:
        matrice_A.append([x, 1])
        
    # On prépare le vecteur B avec les valeurs de y
    vecteur_B = [p[1] for p in points]
    
    # On convertit les listes en tableaux NumPy
    A = np.array(matrice_A)
    B = np.array(vecteur_B)
    
    # On résout le système d'équations linéaires pour trouver les coefficients [a, b, c, d]
    coefficients = np.linalg.solve(A, B)
    
    return coefficients
```

```python
minuscules = [_ for _ in range (ord('a'),ord('z')+1)]
majuscules = [_ for _ in range (ord('A'),ord('Z')+1)]
chiffres = [_ for _ in range (ord('0'),ord('9')+1)]
caracteres_speciaux = [ord (_) for _ in "&~!$#{}[]()<>=+-*%:;,?|_^@/"]

symboles = [0] + minuscules + majuscules + chiffres + caracteres_speciaux

def encode(chaine): return [symboles.index(ord(c)) for c in chaine]

def decode(chaine): return "".join([chr(symboles[v]) for v in chaine])

def gadr(adr):
    #adr = encode(''.join(theAdr.split(':')[::-1]))
    s=0
    i=1
    for c in adr:
        s = s + i*c
        i=i+1
        yield s %89
```
